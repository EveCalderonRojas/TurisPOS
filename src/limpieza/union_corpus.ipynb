# UNIÓN DE CORPUS PREVIOS

Unimos los dos corpus que tenemos:
- Corpus del proyecto 2
- Corpus con las nuevas reseñas

```python
# Importación de librerías
import pandas as pd
```

Leemos los corpus


```python
corpus1 = pd.read_csv("../../data/raw/corpus.csv")
corpus1.head()
```

```python
corpus2 = pd.read_csv("../../data/raw/corpus2.csv")
corpus2.head()
```

Nos quedamos solo con las columnas necesarias para el modelo posterior


Guardamos el corpus nuevo en su lugar respectivo

```python
columnas_base = ['lugar', 'texto', 'calificacion', 'categoria', 'fuente',
                 'idioma_comentario', 'comentarios_espanol', 'emojis']
corpus1 = corpus1[columnas_base]
corpus2 = corpus2[columnas_base]

# Unir
corpus_final = pd.concat([corpus1, corpus2], ignore_index=True)

# Agregar polaridad
def clasificar_polaridad(cal):
    if cal in [1, 2]: return 'negativo'
    if cal == 3:      return 'neutro'
    if cal in [4, 5]: return 'positivo'
    return 'sin_clasificar'

corpus_final['polaridad'] = corpus_final['calificacion'].apply(clasificar_polaridad)

# Guardar
corpus_final.to_csv('../../data/processed/corpus_final.csv', index=False)

print(f'Total filas : {len(corpus_final)}')
print(f'Columnas    : {corpus_final.columns.tolist()}')
print()
print('Distribución polaridad:')
print(corpus_final['polaridad'].value_counts())
print()
print('Distribución categoría:')
print(corpus_final['categoria'].value_counts())
```
