## 1. Inicializar TuriSito

```python
API_KEY = os.getenv('GEMINI_API_KEY', 'XXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXX')
turisito = TuriSito(api_key_gemini=API_KEY)
```

## 2. Función auxiliar para conversar

```python
def conversar(bot, pregunta, mostrar_debug=False):
    resultado = bot.responder(pregunta)
    print(f'Usuario  : {pregunta}')
    print(f'TuriSito : {resultado["respuesta"]}')
    if mostrar_debug:
        print(f'  [Debug] Categoria : {resultado["categoria_detectada"]}')
        print(f'  [Debug] Polaridad : {resultado["polaridad_detectada"]}')
        print(f'  [Debug] Confianza : {resultado["confianza_clf"]:.2%}')
        print(f'  [Debug] Chunks    : {resultado["chunks_usados"]}')
    print('-' * 70)
```

## 3. Pruebas de conversación

```python
# Prueba 1: lugares de naturaleza
conversar(turisito, '¿Qué lugares de naturaleza me recomendás en Costa Rica?', mostrar_debug=True)
```

```python
# Prueba 2: seguimiento (memoria conversacional)
conversar(turisito, '¿Y cuál de esos tiene mejor acceso para familias con niños?')
```

```python
# Prueba 3: restaurantes
conversar(turisito, '¿Qué restaurantes tienen buenas reseñas de comida?', mostrar_debug=True)
```

```python
# Prueba 5: reseñas negativas
conversar(turisito, '¿Hay algún lugar con quejas frecuentes en las reseñas?', mostrar_debug=True)
```

## 4. Ver historial completo

```python
historial = turisito.get_historial()
print(f'Total turnos: {len(historial) // 2}\n')
for turno in historial:
    rol = '👤 Usuario  ' if turno['role'] == 'user' else '🌿 TuriSito '
    contenido = turno['content']
    print(f'{rol}: {contenido[:100]}...' if len(contenido) > 100 else f'{rol}: {contenido}')
    print()
```

## 5. Nueva conversación desde cero

```python
turisito.limpiar_historial()

conversar(turisito, 'Hola TuriSito, ¿qué lugares me recomendás para ecoturismo?')
conversar(turisito, '¿Cuál tiene mejor calificación?')
conversar(turisito, '¿Y qué dicen de la atención de los guías en ese lugar?')
```
